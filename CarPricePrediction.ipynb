```python
from sklearn.preprocessing import OneHotEncoder,LabelEncoder,OrdinalEncoder,StandardScaler
from sklearn.model_selection import train_test_split
X = df.drop('Price',axis=1)
y = df['Price']
```

```python
X.head()

```

```python
y.head()
```

```python
oe = OneHotEncoder(sparse_output=False,drop='first',handle_unknown='ignore')
brand_Fuel_encoded = oe.fit_transform(df[['Brand','Fuel_Type']])
brand_Fuel_encoded_df = pd.DataFrame(brand_Fuel_encoded,columns=oe.get_feature_names_out(['Brand','Fuel_Type']))
brand_Fuel_encoded_df.head()
```

```python
X=pd.concat([X.drop(['Brand','Fuel_Type'],axis=1),brand_Fuel_encoded_df],axis=1)
X.head()
```

```python
X.Year = 2026-X.Year
X.head()
```

```python
label_encoder = LabelEncoder()
X['Transmission'] = label_encoder.fit_transform(X['Transmission'])
X.head()
```

```python
ordinal = OrdinalEncoder(categories=[['First','Second','Third']])
X[['Owner']] = ordinal.fit_transform(X[['Owner']])
X.head()
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.30,random_state=33)
```

```python
stdSc = StandardScaler()
X_train_scaled = stdSc.fit_transform(X_train)
X_test_scaled = stdSc.transform(X_test)
```

```python
np.array(X_train_scaled).shape
```

```python
X_train_scaled.shape
```

```python
X_test_scaled.shape
import pickle
```

```python
with open('models/one_hot_encoder.pkl','wb') as file:
    pickle.dump(oe,file)
with open('models/label_encoder.pkl','wb') as file:
    pickle.dump(label_encoder,file)
with open('models/ordinal_encoder.pkl','wb') as file:
    pickle.dump(ordinal,file)
with open('models/StandardScaler.pkl','wb') as file:
    pickle.dump(stdSc,file)
```

```python
X.head()
```

```python
model = Sequential()
model.add(Dense(256,activation='relu',input_shape=(X_train_scaled.shape[1],)))
model.add(Dense(128,activation='relu'))
model.add(Dense(64,activation='relu'))
model.add(Dense(32,activation='relu'))
model.add(Dense(1))

model.compile(optimizer='adam',loss='mse',metrics=['mae'])
model.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping
early_stopping_calback = EarlyStopping(monitor='val_loss',patience=10,restore_best_weights=True)
```

```python
history = model.fit(
    X_train_scaled,
    y_train,
    validation_split=0.3,
    epochs=200,
    batch_size = 32,
    callbacks=[early_stopping_calback]
)


from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

y_pred = model.predict(X_test_scaled).flatten()

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R2  :", r2)
```
